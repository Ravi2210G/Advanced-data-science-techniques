```python
import pandas as pd
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
data= { 'Age': [ 25,45,31,52,22]
      }
df=pd.DataFrame(data)
print(df)
```

```python
scaler = MinMaxScaler()
```

```python
scaled = scaler.fit_transform(df[['Age']])
```

```python
df_scaled= pd.DataFrame(scaled, columns=['Age_scaled'])
```

```python
print(df_scaled)
```
